```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
df = pd.read_csv(
    "../data/processed_hospital_data.csv"
)

print("Dataset Shape:", df.shape)

df.head()
```

```python
X = df.drop(
    columns=["Waiting_Time_Minutes"]
)

y = df["Waiting_Time_Minutes"]

print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
linear_model = LinearRegression()

linear_model.fit(
    X_train,
    y_train
)

print("Linear Regression model trained successfully!")
```

```python
y_pred_linear = linear_model.predict(
    X_test
)

print("First 10 predictions:")
print(y_pred_linear[:10])
```

```python
mae = mean_absolute_error(
    y_test,
    y_pred_linear
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_linear
    )
)

r2 = r2_score(
    y_test,
    y_pred_linear
)

print("Linear Regression Results")
print("-------------------------")
print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 2))
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
import pandas as pd

df_crowd = pd.read_csv("../data/hospital_data.csv")

print("Dataset loaded successfully!")
print("Shape:", df_crowd.shape)
print(df_crowd.head())
```

```python
def classify_crowd(queue):
    if queue <= 8:
        return "Low"
    elif queue <= 15:
        return "Moderate"
    else:
        return "High"


df_crowd["Crowd_Level"] = df_crowd["Queue_Length"].apply(classify_crowd)

print("Crowd level created successfully!")

print("\nCrowd Level Distribution:")
print(df_crowd["Crowd_Level"].value_counts())

print("\nFirst 10 records:")
print(df_crowd[["Queue_Length", "Crowd_Level"]].head(10))
```

```python
print("Crowd Level Counts:")
print(df_crowd["Crowd_Level"].value_counts())

print("\nCrowd Level Percentages:")
print(
    (df_crowd["Crowd_Level"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
from sklearn.model_selection import train_test_split

X_crowd = df_crowd[
    [
        "Queue_Length",
        "Doctors_Available",
        "Appointments_Count",
        "Consultation_Duration_Min"
    ]
]

y_crowd = df_crowd["Crowd_Level"]

print("X Shape:", X_crowd.shape)
print("y Shape:", y_crowd.shape)
```

```python
X_train_crowd, X_test_crowd, y_train_crowd, y_test_crowd = train_test_split(
    X_crowd,
    y_crowd,
    test_size=0.20,
    random_state=42,
    stratify=y_crowd
)

print("Training samples:", len(X_train_crowd))
print("Testing samples:", len(X_test_crowd))

print("\nTraining class distribution:")
print(y_train_crowd.value_counts())

print("\nTesting class distribution:")
print(y_test_crowd.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(
    X_train_crowd,
    y_train_crowd
)

print("Logistic Regression model trained successfully!")
```

```python
y_pred_logistic = logistic_model.predict(X_test_crowd)

print("First 20 Logistic Regression predictions:")
print(y_pred_logistic[:20])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

logistic_accuracy = accuracy_score(
    y_test_crowd,
    y_pred_logistic
)

logistic_precision = precision_score(
    y_test_crowd,
    y_pred_logistic,
    average="weighted"
)

logistic_recall = recall_score(
    y_test_crowd,
    y_pred_logistic,
    average="weighted"
)

logistic_f1 = f1_score(
    y_test_crowd,
    y_pred_logistic,
    average="weighted"
)

print("Logistic Regression Results")
print("---------------------------")
print("Accuracy :", round(logistic_accuracy, 2))
print("Precision:", round(logistic_precision, 2))
print("Recall   :", round(logistic_recall, 2))
print("F1-Score :", round(logistic_f1, 2))

print("\nClassification Report:")
print(classification_report(y_test_crowd, y_pred_logistic))

print("Confusion Matrix:")
print(confusion_matrix(y_test_crowd, y_pred_logistic))
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_classifier = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

tree_classifier.fit(
    X_train_crowd,
    y_train_crowd
)

print("Decision Tree Classifier trained successfully!")
```

```python
y_pred_tree_crowd = tree_classifier.predict(X_test_crowd)

print("First 20 Decision Tree predictions:")
print(y_pred_tree_crowd[:20])
```

```python
tree_accuracy = accuracy_score(
    y_test_crowd,
    y_pred_tree_crowd
)

tree_precision = precision_score(
    y_test_crowd,
    y_pred_tree_crowd,
    average="weighted"
)

tree_recall = recall_score(
    y_test_crowd,
    y_pred_tree_crowd,
    average="weighted"
)

tree_f1 = f1_score(
    y_test_crowd,
    y_pred_tree_crowd,
    average="weighted"
)

print("Decision Tree Classifier Results")
print("--------------------------------")
print("Accuracy :", round(tree_accuracy, 2))
print("Precision:", round(tree_precision, 2))
print("Recall   :", round(tree_recall, 2))
print("F1-Score :", round(tree_f1, 2))

print("\nClassification Report:")
print(classification_report(
    y_test_crowd,
    y_pred_tree_crowd
))

print("Confusion Matrix:")
print(confusion_matrix(
    y_test_crowd,
    y_pred_tree_crowd
))
```
